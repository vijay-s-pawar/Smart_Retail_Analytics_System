```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# ===============================
# 📦 CLASS (OOP)
# ===============================
class RetailAnalytics:

    def __init__(self, file_path):
        self.file_path = file_path
        self.df = None

    # ===============================
    # 🔹 1. DATA LOADING
    # ===============================
    def load_data(self):                           # 4. then, load csv file as a data frame with handle error
        try:
            self.df = pd.read_csv(self.file_path)   
            print("✅ Data Loaded Successfully")
        except FileNotFoundError:
            print("❌ File not found")
        except Exception as e:
            print("❌ Error:", e)

    # ===============================
    # 🔹 2. DATA CLEANING
    # ===============================
    def clean_data(self):                          # 6. then, clean data : empty customer → Unknown 
        df = self.df                                                     # Mumbai  → mumbai
                                                                         # Invalid date → becomes NULL
        # Handle missing values                                          # Text → number, Invalid → NULL
        df['Customer'].fillna('Unknown', inplace=True)                   # Removes rows with NULL values

        # Standardize city names
        df['City'] = df['City'].str.strip().str.lower()

        # Convert Date
        df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

        # Convert Price
        df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

        # Remove invalid rows
        df.dropna(inplace=True)

        self.df = df
        print("✅ Data Cleaned")

    # ===============================
    # 🔹 3. FEATURE ENGINEERING
    # ===============================
    def feature_engineering(self):                 # 8. New column created: Revenue = Quantity × Price
        self.df['Revenue'] = self.df['Quantity'] * self.df['Price']
        print("✅ Feature Created (Revenue)")

    # ===============================
    # 🔹 4. EDA ANALYSIS
    # ===============================
    def analyze(self):                             # 10. then get: sum of total revenue
        df = self.df                                             # group by and sum of total revenue
                                                                 # Removes duplicates 
        self.total_revenue = df['Revenue'].sum()

        self.top_products = df.groupby('Product')['Revenue'].sum().sort_values(ascending=False).head(5)

        self.city_sales = df.groupby('City')['Revenue'].sum().sort_values(ascending=False)

        self.monthly_sales = df.groupby(df['Date'].dt.to_period('M'))['Revenue'].sum()

        self.category_perf = df.groupby('Category')['Revenue'].sum()

        # Dictionary → customer spending
        self.customer_spending = df.groupby('Customer')['Revenue'].sum().to_dict()

        # Set → unique cities
        self.unique_cities = set(df['City'])

        print("✅ Analysis Completed")

    # ===============================
    # 🔹 5. VISUALIZATION
    # ===============================
    def visualize(self):                       # 12. then, plot chart in x & y asix and save in folder and reset(clean) for next chart
        os.makedirs("charts", exist_ok=True)

        # 1. Top 5 Products
        self.top_products.plot(kind='bar')
        plt.title("Top 5 Products")
        plt.xlabel("Product")
        plt.ylabel("Revenue")
        plt.savefig("charts/top_products.png")
        plt.clf()

        # 2. City-wise Sales
        self.city_sales.head(10).plot(kind='bar')
        plt.title("City-wise Sales")
        plt.xlabel("City")
        plt.ylabel("Revenue")
        plt.savefig("charts/city_sales.png")
        plt.clf()

        # 3. Monthly Sales Trend
        self.monthly_sales.plot(kind='line')
        plt.title("Monthly Sales Trend")
        plt.xlabel("Month")
        plt.ylabel("Revenue")
        plt.savefig("charts/monthly_sales.png")
        plt.clf()

        # 4. Category Contribution
        self.category_perf.plot(kind='pie', autopct='%1.1f%%')
        plt.title("Category Contribution")
        plt.ylabel("")
        plt.savefig("charts/category.png")
        plt.clf()

        print("✅ Charts Created")

    # ===============================
    # 🔹 6. REPORT GENERATION
    # ===============================
    def generate_report(self):                              # 14. open and write new file 
        top_customer = sorted(self.customer_spending.items(), key=lambda x: x[1], reverse=True)[:5]

        with open("final_report.txt", "w") as f:
            f.write(f"Total Revenue: {self.total_revenue}\n")
            f.write(f"Top Product: {self.top_products.idxmax()}\n")
            f.write(f"Top City: {self.city_sales.idxmax()}\n")
            f.write(f"Top Category: {self.category_perf.idxmax()}\n")

            f.write("\nTop 5 Customers:\n")
            for cust, val in top_customer:
                f.write(f"{cust}: {val}\n")

            f.write("\nUnique Cities:\n")
            for city in self.unique_cities:
                f.write(f"{city}\n")

        print("✅ Report Generated")


# ===============================
# 🚀 MAIN EXECUTION
# ===============================
if __name__ == "__main__":                               # 1. “Start running the program from here”

    obj = RetailAnalytics("raw_retail_dataset_500.csv")  # 2. A class object is created and call internally (empty data internally) to RetailAnalytics() function 

    obj.load_data()                                      # 3. this object call to load_data() function
    obj.clean_data()                                     # 5. this object call to clean_data() function
    obj.feature_engineering()                            # 7. this object call to feature_engineering() function
    obj.analyze()                                        # 9. this object call to analyze() function
    obj.visualize()                                      # 11. this object call to visualize() function
    obj.generate_report()                                # 13. this object call to generate_report() function

```
